# Deployment dan Monitoring
<p>Setelah model machine learning dilatih, diuji, dan dioptimalkan, langkah berikutnya adalah deployment (penerapan) dan monitoring (pemantauan) dari model tersebut. Hal ini menjadi penting dikarenakan tahapan ini satu-satunya cara agar model yang Anda bangun sebelumnya dapat dikonsumsi oleh masyarakat umum.</p>

### Apa itu Deployment?
<p>Deployment adalah proses pelatihan model dengan mengintegrasikannya ke dalam aplikasi atau sistem produksi, sehingga dapat digunakan oleh pengguna akhir (end user). Monitoring adalah proses untuk memastikan bahwa model bekerja dengan baik setelah deployment dan untuk mengidentifikasi jika ada degradasi kinerja atau masalah lain.</p>
<p>Pada latihan sebelumnya, kita memiliki sebuah model regresi yang disimpan menjadi dua ekstensi yang berbeda yaitu joblib dan pickle. Langkah pertama untuk melakukan deployment adalah memuat model yang sudah kita bangun.</p>

```python
import pickle
import joblib

# Memuat model dari file joblib
joblib_model = joblib.load('flask_deployment/gbr_model.joblib')

# Membuat model dari file pickle
with open('flask_deployment/gbr_model.pkl', 'rb') as file:
    pickle_model = pickle.load(file)
```

<p>Setelah model dimuat, langkah berikutnya adalah mengintegrasikan model tersebut ke dalam aplikasi produksi. Ini bisa dilakukan dalam berbagai konteks, seperti web application, API, atau edge devices.</p>
<p>Pada contoh kasus modul ini, Anda akan berlatih untuk membuat sebuah API menggunakan Flask sehingga bisa digunakan oleh user melalui API.</p>

<code>
<p style="margin-bottom:-1.5%;">from flask import Flask, request, jsonify</p>
<p>import joblib<p>
 
<p style="margin-bottom:-1.5%;"># Inisialisasi aplikasi Flask</p>
<p>app = Flask(__name__)</p>
 
<p style="margin-bottom:-1.5%;"># Memuat model yang telah disimpan</p>
<p>joblib_model = joblib.load('gbr_model.joblib') # Pastikan path file sesuai dengan penyimpanan Anda</p>
 
<p style="margin-bottom:-1.5%;">@app.route('/predict', methods=['POST'])</p>
<p style="margin-bottom:-1.5%;">def predict():</p>
<p style="margin-bottom:-1.5%;">&nbsp;&nbsp;data = request.json['data']  # Mengambil data dari request JSON</p>
<p style="margin-bottom:-1.5%;">&nbsp;&nbsp;prediction = joblib_model.predict(data)  # Melakukan prediksi (harus dalam bentuk 2D array)</p>
<p>&nbsp;&nbsp;return jsonify({'prediction': prediction.tolist()})</p>
 
<p style="margin-bottom:-1.5%;">if __name__ == '__main__':</p>
<p>&nbsp;&nbsp;app.run(debug=True)</p>
</code>
